# Environment and selected dataset

Check the runtime, resolved configuration and enabled input schemas before spending time on a data run.

Select the dataset through the Compose environment or set `MEDLENS_DATASET` before opening the kernel. The resolved configuration determines the enabled tables and policies.

```python
%matplotlib inline
from medlens.notebook import workspace, notebook_run
import matplotlib.pyplot as plt

research = workspace()
```

```python
print("Dataset", research.config.dataset, research.config.version)
print("Configuration SHA-256", research.config_hash)
print("Spark", research.spark.version, "SQL probe", research.spark.sql("SELECT 1 AS ok").first().ok)
schemas = {name: len(spec.columns) for name, spec in research.config.sources.items()}
plt.barh(list(schemas), list(schemas.values()))
plt.xlabel("Declared source fields")
plt.tight_layout()
plt.show()
```
