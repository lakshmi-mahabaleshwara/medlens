# Map source evidence into canonical clinical entities

Inspect output cardinality and canonical validation issues. Each canonical row retains its original source occurrence and mapping version.

Select the dataset through the Compose environment or set `MEDLENS_DATASET` before opening the kernel. The resolved configuration determines the enabled tables and policies.

```python
%matplotlib inline
from medlens.notebook import workspace, notebook_run
import matplotlib.pyplot as plt

research = workspace()
```

The run records its input versions, resolved configuration, and producing stages. Set `MEDLENS_PARENT` to replay from a specific published manifest. The default walkthrough uses the preceding notebook checkpoint for this configuration.

```python
run = notebook_run(research, 'standardization')
print("Run", run["run_id"], "Parent", run["parent"])
```

```python
mapped = {k: v for k, v in run["tables"].items() if v["role"] == "canonical"}
names = list(mapped)
plt.barh(names, [mapped[k]["stats"]["rows"] for k in names])
plt.xlabel("Canonical occurrences")
plt.tight_layout()
plt.show()
for stage in research.config.stages.values():
    if stage.enabled and stage.op == "validate" and stage.inputs[0] in mapped:
        print(stage.subject, run["tables"][stage.output]["stats"])
```
