# Register an immutable source delivery

Register the enabled files, verify their byte checksums and inspect the malformed-record accounting. The registered references can be replayed from another kernel.

Select the dataset through the Compose environment or set `MEDLENS_DATASET` before opening the kernel. The resolved configuration determines the enabled tables and policies.

```python
%matplotlib inline
from medlens.notebook import workspace, notebook_run
import matplotlib.pyplot as plt

research = workspace()
```

The run records its input versions, resolved configuration, and producing stages. Set `MEDLENS_PARENT` to replay from a specific published manifest. The default walkthrough uses the preceding notebook checkpoint for this configuration.

```python
run = notebook_run(research, 'ingestion')
print("Run", run["run_id"], "Parent", run["parent"])
```

```python
sources = {k: v for k, v in run["tables"].items() if v["role"] == "source"}
for name, ref in sources.items():
    print(name, ref["version"], ref["stats"])
plt.barh(list(sources), [v["stats"]["rows"] for v in sources.values()], label="Source rows")
plt.barh(list(sources), [v["stats"]["malformed"] for v in sources.values()], label="Malformed")
plt.xlabel("Logical CSV records")
plt.legend()
plt.tight_layout()
plt.show()
```
